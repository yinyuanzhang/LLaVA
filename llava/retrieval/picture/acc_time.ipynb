```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 设置 seaborn 的风格为白色背景，以便更好地控制绘图细节
sns.set_style("whitegrid")

# 准备数据 (假设你已经定义了 df, method_colors, method_markers)
data = {
    'model': ['Mistral-7B'] * 4,
    'dataset': ['2WikiMQA'] * 4,
    'method': ['CacheBlend', 'Full KV reuse', 'Prefix Caching', 'Full KV recompute'],
    'TTFT': [0.2, 0.4, 0.6, 0.8],
    'score': [0.3, 0.2, 0.1, 0.05],
}
df = pd.DataFrame(data)

method_colors = {
    'CacheBlend': 'red',
    'Full KV reuse': 'brown',
    'Prefix Caching': 'blue',
    'Full KV recompute': 'green',
}

method_markers = {
    'CacheBlend': 's',
    'Full KV reuse': 'x',
    'Prefix Caching': 'o',
    'Full KV recompute': '^',
}

# 创建子图布局
fig, ax = plt.subplots(figsize=(6, 5))

# 绘制散点图
for method in df['method'].unique():
    subset = df[df['method'] == method]
    ax.scatter(subset['TTFT'], subset['score'], 
               color=method_colors[method], 
               marker=method_markers[method], 
               s=100, 
               label=method)

# 设置图表标题和坐标轴标签
ax.set_title('Mistral-7B', fontsize=12) 
ax.set_xlabel('TTFT (s)', fontsize=10)
ax.set_ylabel('F1-score', fontsize=10)

# 设置坐标轴范围
ax.set_xlim(0, 1)  
ax.set_ylim(0, 0.4) 

# 在图表上添加 "Better" 箭头和文字注解
ax.annotate('Better', xy=(0.75, 0.28), xytext=(0.55, 0.28),
            arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=8),
            fontsize=12, ha='center', va='center')

# 添加图例
ax.legend(loc='lower right', bbox_to_anchor=(1.2, 0))

# 调整布局，防止标签重叠
plt.tight_layout()

# 保存图表为 PDF 文件
plt.savefig('acc_time.pdf', bbox_inches='tight')

# 显示图表（可选，如果你想在 Jupyter Notebook 中看到它）
plt.show()
```
